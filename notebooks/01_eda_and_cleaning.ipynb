```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/dataset.csv")
```

```python
df
```

```python
df.info()
```

```python
print("raw:", df.shape)
```

```python
df = df.drop(columns=["Unnamed: 0"])
```

```python
df
```

```python
df = df.dropna()
```

```python
# 4. remove duplicate songs
df = df.drop_duplicates(subset="track_id")
print("after track_id dedupe: ",df.shape)
df = df.drop_duplicates(subset=["track_name","artists"])
```

```python
# 5. remove outliers
df = df[(df["duration_ms"]>=60000) & (df["duration_ms"]<=600000)]
df = df[df["tempo"]>0]
df = df.reset_index(drop=True)
print("clean:",df.shape)
```

```python
# save the clean data (data/processed is ignored by git)
df.to_csv("../data/processed/clean_tracks.csv", index=False)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
features = ["danceability", "energy", "loudness", "speechiness",
            "acousticness", "instrumentalness", "liveness",
            "valence", "tempo"]
```

```python
df[features].hist(figsize=(12,8),bins=30)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df[features].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.show()
```
